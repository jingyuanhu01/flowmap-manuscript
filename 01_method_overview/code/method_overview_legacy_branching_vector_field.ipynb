```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Random seed for reproducibility
np.random.seed(42)

# Generate small fake matrices
expression_matrix = np.random.normal(loc=0, scale=1, size=(5, 6))
velocity_matrix = np.random.normal(loc=0, scale=0.5, size=(5, 6))

# Plot both heatmaps with different colormaps and no text
fig, axes = plt.subplots(1, 2, figsize=(6, 3))

sns.heatmap(expression_matrix, ax=axes[0], cmap="viridis", cbar=False, xticklabels=False, yticklabels=False)
axes[0].set_xticks([])
axes[0].set_yticks([])

sns.heatmap(velocity_matrix, ax=axes[1], cmap="magma", cbar=False, xticklabels=False, yticklabels=False)
axes[1].set_xticks([])
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sys
sys.path.append("..")

from scripts.TPS import ThinPlateSpline
from scripts.plotting import plot_velocity_streamplot

# Load branch_4 vector field
df = pd.read_csv("./data/1d/branch_4.csv")
X = df[["x", "y"]].values
V = df[["vx", "vy"]].values
time = df["time"].values
time = (time - np.min(time)) / (np.max(time) - np.min(time))  # normalize for color map

# Fit Thin Plate Spline to vector field
tps_vf = ThinPlateSpline(X, n_control_points=100)
tps_vf.fit(V, dof=15)

# Plot schematic
fig, ax = plt.subplots(figsize=(5, 5))

plot_velocity_streamplot(
    X_2d=X,
    tps_vf=tps_vf,
    grid_density=1.0,
    stream_density=1.0,
    scatter_color=time,  # keep colorful scatter
    scatter_size=200,
    scatter_alpha=0.1,
    ax=ax,
    title="",
    figsize=(5, 5),
    aspect="equal",
    vmin=0.0,
    vmax=1.0,
    grid_size=50
)

# Remove all visual clutter
ax.set_xticks([])
ax.set_yticks([])
for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```
