```python
import numpy as np
import anndata as ad
import dynamo as dyn

# --------------------------
# 0. Paths
# --------------------------
outdir = "./data/benchmark/velocity_estimation/dentate_gyrus/"
S_path = f"{outdir}/spliced.npy"
U_path = f"{outdir}/unspliced.npy"

# --------------------------
# 1. Load matrices
# --------------------------
S = np.load(S_path)
U = np.load(U_path)

print("Loaded S, U:", S.shape, U.shape)

# --------------------------
# 2. Build AnnData (minimal)
# --------------------------
adata = ad.AnnData(X=S)
adata.layers["spliced"] = S
adata.layers["unspliced"] = U

# --------------------------
# 3. Preprocessing (minimal)
# --------------------------
dyn.pp.recipe_monocle(adata)

# --------------------------
# 4. Compute velocity (core step)
# --------------------------
dyn.tl.dynamics(adata)

# --------------------------
# 5. Extract velocity matrix
# --------------------------
# this is the key line
V_dyn = adata.layers["velocity_S"]

if not isinstance(V_dyn, np.ndarray):
    V_dyn = V_dyn.toarray()

np.save(f"{outdir}/V_dynamo.npy", V_dyn)
print("Saved Dynamo velocity:", V_dyn.shape)

print("\n✅ Done: Dynamo raw velocity saved")
```
