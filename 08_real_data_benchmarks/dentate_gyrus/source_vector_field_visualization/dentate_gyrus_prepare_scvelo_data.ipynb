```python
import scvelo as scv
import numpy as np
import pandas as pd
import os

# Config
scv.settings.verbosity = 3
scv.settings.presenter_view = True
scv.settings.set_figure_params("scvelo")

# Load dataset
adata = scv.datasets.dentategyrus()

# Preprocessing
scv.pp.filter_and_normalize(adata, min_shared_counts=30, n_top_genes=2000)
scv.pp.moments(adata, n_pcs=30, n_neighbors=30)

# Velocity estimation
scv.tl.velocity(adata)
scv.tl.velocity_graph(adata)

# Plot velocity stream
scv.pl.velocity_embedding_stream(adata, basis="umap")
```

```python
scv.tl.recover_dynamics(adata)
```

```python
scv.tl.velocity(adata, mode='dynamical', vkey='dynamical_velocity')
scv.tl.velocity_graph(adata, vkey='dynamical_velocity', variance_stabilization=True)
scv.pl.velocity_embedding_stream(adata, vkey='dynamical_velocity')
```

```python
# Save data outputs
adata.write("./data/dentate_gyrus/dentategyrus_velocity_processed.h5ad", compression="gzip")
```
