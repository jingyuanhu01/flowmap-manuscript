```python
%load_ext autoreload
%autoreload 2

import numpy as np
import scanpy as sc
import scvelo as scv
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler
from scipy.sparse import issparse
import flowmap
from flowmap import *

# --------------------------
# 0. Load
# --------------------------
adata = sc.read_h5ad("./data/B_cell/bcell_velocity_standard.h5ad")

# subset to HVGs
hvg_mask = adata.var["highly_variable"].values
X = adata.X[:, hvg_mask]
if issparse(X):
    X = X.toarray()

# V
V = adata.layers["velocity"][:, hvg_mask]
if issparse(V):
    V = V.toarray()

pseudotime = list(adata.obs["velocity_pseudotime"])
```

```python
import umap
import numpy as np

# reproducibility
np.random.seed(0)

X_pca = adata.obsm["X_pca"]

umap_model = umap.UMAP(
    n_neighbors=30,
    n_components=2,
    min_dist=0.5,
    spread=1.0,
    metric="euclidean",
    init="spectral",
    random_state=0
)

X_umap = umap_model.fit_transform(X_pca)

# store back into AnnData (optional)
adata.obsm["X_umap_repro"] = X_umap

import matplotlib.pyplot as plt


plt.scatter(
    adata.obsm["X_umap_repro"][:,0],
    adata.obsm["X_umap_repro"][:,1],
    s=1,
    label="umap pkg"
)

plt.legend()
plt.show()
```

```python
import numpy as np
from sklearn.decomposition import TruncatedSVD

# --------------------------
# 1. Get X (must match Scanpy input)
# --------------------------
X = adata.X.copy()

# convert to dense if needed
if hasattr(X, "toarray"):
    X = X.toarray()

# --------------------------
# 2. Center (Scanpy does this)
# --------------------------
X_mean = X.mean(axis=0)
X_centered = X - X_mean

# --------------------------
# 3. PCA via SVD
# --------------------------
n_pcs = 30

svd = TruncatedSVD(n_components=n_pcs, random_state=0)
X_pca_repro = svd.fit_transform(X_centered)

# --------------------------
# 4. Compare with Scanpy PCA
# --------------------------
X_pca_scanpy = adata.obsm["X_pca"]

# correlation check (per component)
corrs = [
    np.corrcoef(X_pca_repro[:, i], X_pca_scanpy[:, i])[0, 1]
    for i in range(n_pcs)
]

print("Mean correlation:", np.mean(np.abs(corrs)))
print("First 5 correlations:", corrs[:5])
```

```python
import flowmap
from flowmap import *

emb = VectorFieldEmbedder(X, V, method="umap", dist_method="euclidean", alpha=0.0,
                          dof=30, knn_k=30, 
                          embed_kwargs={"n_neighbors":30,
                                        "min_dist":0.5})
emb.fit_embedding(seed=0)

fig = flowmap.plot.plot_velocity_grid(
    emb.X_emb,
    spline_vf=emb.spline_vf,
    grid_size=30,
    grid_density=1.0,
    smooth=0.5,
    n_neighbors=None,
    min_mass=0.01,
    scatter_color=pseudotime,
    scatter_size=10,
    scatter_alpha=0.1,
    arrow_color="black",
    arrow_alpha=0.9,
    arrow_scale=15.0,
    arrow_width=0.0025,
    figsize=(6, 6),
    cmap="viridis",
    vmin=None,
    vmax=None,
    show_axes=False,
    show_colorbar=False,
)
```
