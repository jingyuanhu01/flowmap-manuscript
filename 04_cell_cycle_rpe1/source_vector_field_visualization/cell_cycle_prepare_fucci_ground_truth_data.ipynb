```python
import numpy as np

# Expression (total RNA)
X = adata.layers["X_total"]
np.save("./data/fucci/X.npy", X)

# Velocity (total RNA)
V = adata.layers["velocity_T"]
np.save("./data/fucci/V.npy", V)

# Metadata for evaluation and coloring
metadata = adata.obs[[
    "Cell_cycle_relativePos",
    "cell_cycle_phase",
    "RFP_log10_corrected",
    "GFP_log10_corrected",
    "time",  # optional: labeling time info
]].copy()
metadata.to_csv("./data/fucci/metadata_ground_truth.csv", index=True)  # keep index to link back to AnnData if needed
```
